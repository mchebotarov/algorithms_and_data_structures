**Description**

**Examples**

* Example 3:

    Input: `root = [10, 9, 8, 7]`

    Output: -1

**Problem 1 Starter Code**

You may assume that `root` is a binary tree that has been built, i.e., you can access a node/leaf by calling `root.left.right.left.right...`, so you don't have to provide a tree-building function to convert an input list to a binary tree as an object of the `TreeNode` class.

```python
from collections import deque

# Definition for a binary tree node.
class TreeNode(object):
    def __init__(self, val = 0, left = None, right = None):
        self.val = val
        self.left = left
        self.right = right


def is_duplicate(root: TreeNode) -> int:
    if root is None:
        return -1

    visitedNodes = set()
    queue = deque()
    queue.append(root)

    while queue:
        node = queue.popleft()

        if node.val in visitedNodes:
            return node.val
        visitedNodes.add(node.val)

        if node.left is not None:
            queue.append(node.left)
        if node.right is not None:
            queue.append(node.right)

    return -1

[1, 10, 2, 3, 10, 12, 12]
root = TreeNode(1);
root.left = TreeNode(10)
root.right = TreeNode(2);
root.left.left = TreeNode(3);
root.left.right = TreeNode(10);
root.right.left = TreeNode(12);
root.right.right = TreeNode(12);
root
print(is_duplicate(root))
  
```

3. Walkthrough one example that your partner made and explain it.

```python
# My partner created this example Example 6: • Input: s = "({[})" • Output: False
#
#There is an open '('. followed by a '{' bracket, and then '['. Immediately after there is a '}' which is a closing bracket and it does
# not match the immediately open bracket of '[' before that, so it returns False
```


4. Copy the solution your partner wrote. 


```python
# Your answer here
def is_valid_brackets(s: str) -> bool:
    stack=[]
    mapping = {")":"(","]":"[","}":"{"}

    for char in s:
        if char in mapping:
            if not stack:
                return False
            top_element = stack.pop()
            if top_element != mapping[char]:
                return False

        else:
            stack.append(char)

    return len(stack) == 0
```

5. Explain why their solution works in your own words.


```python
# The solution checks each character in the input string.  It will push to the stack if it's not in the mapping of closing brackets.  
# If it is in the mapping of closing brackets, then it will check if the top element of the stack matches the value of the key of
# the closing bracket.  If it does, it will continue checking the next characters, otherwise it will return false. If a closing bracket 
# appears and the stack is empty, it will also return false as there is no matching open bracket.
# This works as the stack's top element always holds the most recent unmatched opening bracket that needs to be matched by a closing bracket.
# After the string was processed, the stack must be empty for the input to be considered valid.
```

6. Explain the time and space complexity of their solution in your own words and identify if they found the correct time and space complexities.


```python
# It uses both O(n) time and O(n) space.  For time, it will be O(n) because at worst it needs to check each character in the input 
# string once.
# For O(n) space, because there will be at worst case O(n) elements in the stack if all characters are open brackets.

# Yes, they correctly mentioned both the time and space complexity for this solution.
```
